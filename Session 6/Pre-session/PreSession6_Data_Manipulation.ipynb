## Section 3: Advanced usage of numpy

### NumPy 2D Arrays

NumPy arrays are powerful tools for working with large, multidimensional data. In engineering, these arrays (also known as a matrix) can represent matrices, simulation grids, and more.


### Example 3.1: Creating a 2D Array for a Matrix of Stresses

You can create a 2D NumPy array  using the `numpy.array()` function. Each row in the array represents a different set of data, and each column represents a variable or feature.

Let’s create a 2D array representing stresses (in Pascals) measured at different points in a material during a stress test. The rows represent different test points, and the columns represent stress values in different directions.


````python
import numpy as np

# Create a 2D array (matrix) of stresses in Pascals
stresses = np.array([
    [100, 200, 150],  # Point 1: stress in x, y, z directions
    [120, 220, 140],  # Point 2: stress in x, y, z directions
    [130, 230, 160],  # Point 3: stress in x, y, z directions
    [140, 240, 170]   # Point 4: stress in x, y, z directions
])

print("Stresses (Pa):\n", stresses)
````

---

### Indexing 2D Arrays


You can access a specific row or column by using indexing. Remember that Python uses 0-based indexing.


### Example 3.2: Access the Stress Values for the First Test Point (Row 0)


````python
# Access the first row (stress values for Point 1)
first_row = stresses[0, :]
print("Stress values for Point 1 (in x, y, z directions):", first_row)
````



### Example 3.3: Access the Stress in the Y-Direction for All Points (Column 1)


````python
# Access the second column (stress in the y-direction for all points)
y_stress = stresses[:, 1]
print("Stress in the y-direction for all points:", y_stress)
````

---

### Extracting Subsections of an Array

You can also extract larger subsections of the array using slicing.


### Example 3.4: Extract Stress Values for the First Two Points (Rows 0 and 1)


````python
# Extract the first two rows (stress values for the first two points)
subset = stresses[0:2, :]
print("Stress values for the first two points:\n", subset)
````

## Section 4 - Advanced usage of pandas DataFrames

pandas DataFrames are similar to NumPy 2D arrays, but they offer more flexibility, especially when working with labelled data. You can think of DataFrames as tables with rows and columns, much like Excel sheets, but with far more powerful indexing and manipulation capabilities.



### Example 4.1: Creating a DataFrame of Engineering Measurements (without a .csv file)

Let’s create a DataFrame to represent measurements of displacement (in mm) and velocity (in m/s) at different points in a mechanical system during an experiment.


````python
import pandas as pd

points = ['A', 'B', 'C', 'D']
displacements = [5.2, 6.3, 7.1, 4.9]
velocities = [0.3, 0.35, 0.4, 0.25]

# Create a pandas DataFrame for displacement and velocity measurements
data = {
    'Point': points,
    'Displacement (mm)': displacements,
    'Velocity (m/s)': velocities
}

df = pd.DataFrame(data)
print(df)

# Save the DataFrame to a CSV file
df.to_csv('Displacement_and_velocity.csv', index=False)

# Output a message to confirm
print("CSV file 'Displacement_and_velocity.csv' has been saved.")
````

---

### Example 4.2: Slicing a DataFrame using `iloc`
In Session 4 you saw how to access columns of DataFrames by using their headings in square brackets. You can access rows (or columns), by using `.iloc`, which works in a similiar way to array indexing.


````python
# Access the data for point 'C' (third row)
point_c = df.iloc[2, :]
print("Data for Point C:\n", point_c)
````

````python
# Access the data for the second and third column
point_c = df.iloc[:, 1:3]
print("Data for columns 2 and 3:\n", point_c)
````

---

### Data Manipulation in DataFrames using `loc`

pandas allows you to perform advanced manipulations on your data. You can easily filter rows based on conditions or modify values within the DataFrame using `loc`


### Example 4.3: Filter Points with a Displacement Greater Than 6 mm

````python
# Filter the DataFrame for points with a displacement greater than 6 mm
high_displacement = df[df['Displacement (mm)'] > 6]
print("Points with displacement greater than 6 mm:\n", high_displacement)
````



### Example 4.4: Modify the Velocity of Point 'D' to 0.30 m/s


````python
# Modify the velocity of Point D to 0.30 m/s
df.loc[df['Point'] == 'D', 'Velocity (m/s)'] = 0.30
print("Updated DataFrame:\n", df)
````

### Summary of `.loc` vs `.iloc` in pandas

- **`.loc`**: Label-based indexing, used to access data by row and column labels.
  - Syntax: `df.loc[row_label, column_label]`
  - Example:
    ```python
    df.loc['B', 'Displacement (mm)']  # Accesses the Displacement for Point B
    df.loc['A', :]  # Accesses all data for Point A
    ```

- **`.iloc`**: Position-based indexing, used to access data by row and column index positions.
  - Syntax: `df.iloc[row_index, column_index]`
  - Example:
    ```python
    df.iloc[1, 0]  # Accesses the Displacement (mm) for Point B (second row, first column)
    df.iloc[2, :]  # Accesses all data for Point C (third row)
    ```

- **Key Difference**:
  - `.loc` uses labels (row and column names).
  - `.iloc` uses positional indices (row and column positions).
