## Section 1: Conditional Statements: `if`, `else`, and `elif`
Conditional statements allow you to execute different blocks of code based on a condition.


### Example 1.1: `if` statements

The `if` statement followed by a boolean will execute the code only if the value of the boolean is `True`. So we can use it to do things only under certain conditions, for example, when one number is bigger than another. **Note** that the code after the if statement needs to be indented - use the `tab` key [The key to the left of letter `Q`] to create indents. Although, many modern editors used for writing Python do a lot of this automatically.

````python
# Checking if one number is greater than another
a = 33
b = 200

if b > a:
    print("b is greater than a")
````

### Example 1.2: `else` statements

The `else` statement follows the `if` block with the indent removed (so `else` is vertically under `if`). 

The indented statements following `else` will only run if the condition for `if` was not `True`



````python
# else Example
a = 200
b = 33

if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")
````

The `elif` (read as 'else if') statement follows the `if` block with the indent removed. 

The condition following `elif` is only tested when the `if` condition is not `True`. See this example and test different values for `a` and `b` to see these statements in action:

````python
# elif example

a=200
b=200

if b==a:
    print("a and b are equal")
elif a>b:
    print("a is greater than b")
else:
    print("b is greater than a")
````

### Nested Conditionals
Python allows you to write conditionals inside other conditionals, which is known as nesting. This can be useful when making more complex decisions.

---

### Example 1.3: Nested Conditionals - Check if a number is greater than 10 and whether it is odd or even
In this example, we will determine whether a number is greater than 10 and, if so, whether it is even or odd.


````python
# Example 1.3: Nested conditional statements
x = 13

if x > 10:
    if x % 2 == 0:
        print("Number is greater than 10 and even")
    else:
        print("Number is greater than 10 and odd")
else:
    print("Number is less than 10")
````

## Section 2: Iterative Statements: `for` and `while` Loops
Loops allow us to execute a block of code repeatedly. Python provides two types of loops:

- `while` loop: Repeats a block of code as long as a specified condition is true.
- `for` loop: Iterates over a sequence (such as a list or range) a specific number of times.

---
### `while` Loops
A `while` loop continues to execute as long as the condition remains `True`. Once the condition becomes `False`, the loop stops.

### Syntax:
```python
while condition:
    # code block that runs repeatedly as long as condition is True


````python

# Example 2.1: A `while` loop that adds 1 until the value reaches 10
N = 10
n = 0

while n < N:
    n = n + 1
    print(n)
````

### `for` Loops
A `for` loop iterates over a sequence (such as a list, tuple, or range). The loop runs for each item in the sequence.

### Syntax:
```python
for variable in sequence:
    # code block executed for each element in the sequence


````python

# Example 2.2: A `for` loop to print items from a list
fruit = ["apple", "pear", "banana", "cherry"]

for x in fruit:
    print(x)
````

### Using `range()` in `for` Loops
The `range()` function generates a sequence of numbers, which is useful for specifying the number of times a loop should run. `range(x,y)` generates a sequence from x to y-1.



### Example 2.1: Using `range()` in `for` loops
Here is an example of a `for` loop that prints the squares of numbers from 0 to 4.


````python
# Example 2.3: Using `range` to iterate and print square numbers
for n in range(5):
    m = n**2
    print(m)
````
